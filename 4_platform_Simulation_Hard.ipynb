## 4-Platform Chain Sharing UCB — Hard CTR Setting

**Setting:** K=4 platforms, 2 arms each, chain sharing (platform i borrows from i−1).  
Platforms ordered by descending gap: delta_1 > ... > delta_4.

**CTR parameters (Hard):**
- `delta_min=0.001`, `delta_max=0.006` — very small arm gaps, hard to identify best arm
- `mu2_range=(0.005, 0.04)` — very low base CTR 0.5%–4%, high noise per observation
- `offset_range=(-0.002, 0.002)` — tighter anchoring, minimises cross-platform bias
- `M0=50`, `K_rep=1000`, `T` tested: 500 / 1000 / 2500 / 5000 / 10000

### Short Run — T = 500

```python
result_4p_hard_500 = run_chain_simulation(
    K=4,
    delta_min=0.001, delta_max=0.006,
    mu2_range=(0.005, 0.04),
    offset_range=(-0.002, 0.002),
    eps=0.001, T=500, K_rep=1000, M0=50, C=2.0,
    rng=np.random.default_rng(2026), verbose=True,
)
format_results(result_4p_hard_500)
```
