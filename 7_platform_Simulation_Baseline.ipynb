## 7-Platform Chain Sharing UCB — Baseline CTR Setting

**Setting:** K=7 platforms, 2 arms each, chain sharing (platform i borrows from i−1).  
Platforms ordered by descending gap: delta_1 > ... > delta_7.

**CTR parameters (Baseline):**
- `delta_min=0.003`, `delta_max=0.020`
- `mu2_range=(0.01, 0.10)` — suboptimal arm CTR 1%–10%
- `offset_range=(-0.005, 0.005)` — anchored mode
- `M0=50`, `K_rep=1000`, `T` tested: 500 / 1000 / 2500 / 5000 / 10000

### Short Run — T = 500

```python
result_7p_baseline_500 = run_chain_simulation(
    K=7,
    delta_min=0.003, delta_max=0.02,
    mu2_range=(0.01, 0.1),
    offset_range=(-0.005, 0.005),
    eps=0.002, T=500, K_rep=1000, M0=50, C=2.0,
    rng=np.random.default_rng(2026), verbose=True,
)
format_results(result_7p_baseline_500)
```
