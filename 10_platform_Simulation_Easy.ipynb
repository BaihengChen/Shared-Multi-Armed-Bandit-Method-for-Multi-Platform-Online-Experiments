## 10-Platform Chain Sharing UCB — Easy CTR Setting

**Setting:** K=10 platforms, 2 arms each, chain sharing (platform i borrows from i−1).  
Platforms ordered by descending gap: delta_1 > ... > delta_10.

**CTR parameters (Easy):**
- `delta_min=0.015`, `delta_max=0.050` — larger arm gaps, arms more distinguishable
- `mu2_range=(0.05, 0.20)` — higher base CTR 5%–20%, more signal per draw
- `offset_range=(-0.005, 0.005)` — anchored mode
- `M0=50`, `K_rep=1000`, `T` tested: 500 / 1000 / 2500 / 5000 / 10000

### Short Run — T = 500

```python
result_10p_easy_500 = run_chain_simulation(
    K=10,
    delta_min=0.015, delta_max=0.05,
    mu2_range=(0.05, 0.2),
    offset_range=(-0.005, 0.005),
    eps=0.002, T=500, K_rep=1000, M0=50, C=2.0,
    rng=np.random.default_rng(2026), verbose=True,
)
format_results(result_10p_easy_500)
```
